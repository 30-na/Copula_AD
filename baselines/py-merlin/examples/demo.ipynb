```python
%matplotlib inline
import numpy as np

import merlin
import merlin.utils as utils
```

### MERLIN on collective anomaly with length 200

as tolerance (shown in light green) we use the same value as when calculating hits for the UCR Anomaly Archive: $max(anomaly\_length, 100)$

```python
TS_NAME = "../tests/fixtures/004_UCR_Anomaly_DISTORTEDBIDMC1_2500_5400_5600.txt"
L_MIN = 170
L_MAX = 190

# 1. load timeseries from file
ts = np.loadtxt(TS_NAME, delimiter=",", dtype=float)

# 2. run MERLIN
discords, distances, lengths = merlin.merlin(ts, lmin=L_MIN, lmax=L_MAX, sanitize=False, timeit=True)

# 3. plot result
name_parts = TS_NAME.split("_")
ground_truth = (int(name_parts[5]), int(name_parts[6].split(".")[0]))
tolerance = max(ground_truth[1] - ground_truth[0], 100)
fig, ax = utils.plot(ts, discords, lengths, anomalies=[ground_truth], name=TS_NAME.split(".")[0], tol=tolerance)
fig.show()
```
